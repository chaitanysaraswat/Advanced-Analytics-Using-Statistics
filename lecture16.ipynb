# Mean Squared Error (MSE)

The error is calculated as:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_{pi}-y_{ai})^2
$$

Where:

- $y_{pi}$ = Predicted Value
- $y_{ai}$ = Actual Value
- $n$ = Number of observations

### Why do we square the error?

If

$$
e = y_{pi}-y_{ai}
$$

then

$$
e^2
$$

is always positive and forms a parabola.

```text
      |
   *  |  *
      |
 *    |    *
------|--------> Error
      |
```

Advantages:

- Removes negative signs.
- Penalizes larger errors more.
- Helps optimization algorithms converge smoothly.

---

# Cost Function

The cost function is the average of all losses.

$$
C=\frac{1}{2n}\sum_{i=1}^{n}(y_{pi}-y_{ai})^2
$$

### Difference Between Loss Function and Cost Function

**Loss Function**

- Error for a single observation.

**Cost Function**

- Average/Sum of losses for all observations.

# Differentiability

A function is differentiable if its derivative exists at every point.

---

## Example 1: Differentiable Function

$$
y=x^2
$$

Graph:

```text
      y
      ↑
   *     *
    *   *
     * *
      *
------+--------------→ x
```

This graph is differentiable everywhere because there are no sharp corners.

---

## Example 2: Non-Differentiable Function

$$
y=|x|
$$

Piecewise representation:

$$
y=
\begin{cases}
x, & x\ge0\\
-x, & x<0
\end{cases}
$$

### Graph

```text
      y
      ↑
      |\
      | \
      |  \
------+---\--------→ x
      |   /
      |  /
      | /
      |/
```

This graph is **not differentiable at $x=0$**.

Reason:

- Left-hand derivative ≠ Right-hand derivative.
- Sharp change in slope at origin.

---

# Overfitting

Overfitting occurs when a model becomes too complex and starts learning noise from the training data.

```text
Good Fit

  *     *
    * *
     *
    * *
  *     *

Overfit

 *\_/\/\_/\/\_/\/\_*
```

Characteristics:

- Very low training error.
- Poor testing performance.
- Poor generalization.

---

# Original Cost Function

$$
C=\frac{1}{2n}\sum_{i=1}^{n}(y_{pi}-y_{ai})^2
$$

The coefficients are chosen such that the cost function is minimized.

---

# Lasso Regression (L1 Regularization)

Penalty Term:

$$
\lambda \sum_{i=1}^{n}|R_i|
$$

where:

- $R_i$ = Regression coefficient
- $\lambda$ = Regularization Hyperparameter

Full Cost Function:

$$
C=
\frac{1}{2n}\sum_{i=1}^{n}(y_{pi}-y_{ai})^2
+\lambda \sum_{i=1}^{n}|R_i|
$$

### Geometry of L1 Penalty

The absolute value function used in Lasso:

$$
y=|x|
$$

```text
      y
      ↑
      |\
      | \
------+--\------→ x
      |  /
      | /
      |/
```

### Key Points

- Uses L1 Regularization.
- Coefficients can become exactly zero.
- Performs Feature Selection.
- Useful when many features are irrelevant.

---

# Ridge Regression (L2 Regularization)

Penalty Term:

$$
\lambda \sum_{i=1}^{n}R_i^2
$$

Full Cost Function:

$$
C=
\frac{1}{2n}\sum_{i=1}^{n}(y_{pi}-y_{ai})^2
+\lambda \sum_{i=1}^{n}R_i^2
$$

### Geometry of L2 Penalty

The squared function used in Ridge:

$$
y=x^2
$$

```text
      y
      ↑
   *     *
    *   *
     * *
      *
------+--------------→ x
```

### Key Points

- Uses L2 Regularization.
- Coefficients shrink toward zero.
- Coefficients never become exactly zero.
- Retains all features.

---

# Why Lasso Uses |x| and Ridge Uses x²

### Lasso

$$
|x|
$$

- Creates sharp corners.
- Can force coefficients exactly to zero.
- Useful for feature selection.

### Ridge

$$
x^2
$$

- Smooth and differentiable.
- Easier optimization.
- Shrinks coefficients without removing them.

---

# Lasso vs Ridge

| Feature | Lasso (L1) | Ridge (L2) |
|----------|------------|------------|
| Penalty | $\sum |R_i|$ | $\sum R_i^2$ |
| Feature Selection | Yes | No |
| Coefficients Become Zero | Yes | No |
| Handles Overfitting | Yes | Yes |
| Best For | Irrelevant Features | Multicollinearity |

---

# Parameters vs Hyperparameters

## Parameters

Learned automatically from data.

Examples:

- Regression coefficients
- Neural Network weights

---

## Hyperparameters

Chosen before training.

Examples:

- Learning Rate
- Epochs
- Number of Trees
- Regularization Parameter ($\lambda$)

---

# Interview Questions

### Q1. Why does Lasso perform feature selection?

Because L1 regularization can make coefficients exactly zero.

### Q2. Why can't Ridge perform feature selection?

Because L2 regularization only shrinks coefficients and does not make them exactly zero.

### Q3. Why is $y=|x|$ not differentiable at $x=0$?

Because left-hand derivative ≠ right-hand derivative.

### Q4. Why is $y=x^2$ differentiable everywhere?

Because the curve is smooth and has no sharp corners.

### Q5. Why does regularization reduce overfitting?

It penalizes large coefficients and reduces model complexity.

---

# Quick Revision

1. MSE = Average Squared Error.
2. Cost Function = Average Loss.
3. $y=x^2$ is differentiable everywhere.
4. $y=|x|$ is not differentiable at $x=0$.
5. Lasso = L1 = Feature Selection.
6. Ridge = L2 = Shrinks Coefficients.
7. $\lambda$ is a Hyperparameter.
8. Regularization reduces Overfitting.
9. Lasso can make coefficients exactly zero.
10. Ridge cannot make coefficients exactly zero.

ONE HOT ENCODING:
- MULTIPLE CATEGORIES
- is used for nominal scale variables
- no logical order
- not useful when no of categories are too large
- regression ,ANN, KNN,SVM

ORDINAL ENCODING:
- this will be used with inherent ranking or ratings
- preserve logical order

FREQUENCY ENCODING:
- k

BINARY ENCODING:
- when there are large no of categories


````python
df = pd.DataFrame({
    'Education':[
        'Graduate',
        'PG',
        'High School',
        'Graduate',
        'PG'
    ]
})

education_map = {
    'High School':1,
    'Graduate':2,
    'PG':3
}

df['Education_Encoded'] = (
    df['Education']
    .map(education_map)
)

print(df)
````

````python
# Frequency encoding
df = pd.DataFrame({
    'City':[
        'Mumbai',
        'Mumbai',
        'Mumbai',
        'Delhi',
        'Delhi',
        'Pune'
    ]
})

freq = df.City.value_counts()

df['City_Frequency'] = (df.City.map(freq))
df
````
