```python
df1=x1.value_counts().reset_index()
```

```python
df1
```

```python
df1['Prop']=df1['count']/sum(df1['count'])
df1['pct']=df1['Prop']*100
df1
```

```python
# what pct of consignment top3 materialid contributed
x2=np.array([4,7,5,2,3,8])
np.cumsum(x2)
df1['cumsum']=np.cumsum(df1['pct'])
df1
```
