# Group By Command

```python
grp1=df.groupby('Location')
print(grp1)
print(grp1.groups)# grouping is one on basis of location
df1=grp1.get_group('MWH-2').reset_index(drop=True)#taking out key value pair that has been created by group by on MWH-2
df2=grp1.get_group('MWH-3').reset_index(drop=True)#taking out key value pair that has been created by group by on MWH-3
df3=pd.concat([df1,df2]).reset_index(drop=True)#combining the two df using concat
df3
```

```python
df5=grp1.Quantity.agg(np.sum).reset_index()
df5
```

```python
grp1.Quantity.agg(np.size)
```

```python
df6=grp1.Quantity.agg(np.mean).reset_index()
df6
```
