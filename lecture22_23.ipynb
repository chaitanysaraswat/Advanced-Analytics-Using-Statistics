# ANN Interview Notes

### Vanishing Gradient

Keywords written:

- Vanishing gradient
- Sigmoid activation
- Chain rule derivative

---

### Why Vanishing Gradient Happens

Simple network:

\[
h=W_1x
\]

\[
\hat y=W_2h
\]

Loss:

\[
L=\frac12(\hat y-y)^2
\]

Forward:

\[
W_1
\rightarrow
h
\rightarrow
\hat y
\rightarrow
L
\]

Backward:

\[
L
\rightarrow
\hat y
\rightarrow
h
\rightarrow
W_1
\]

Gradient:

\[
\frac{dL}{dW_1}
=
\frac{dL}{d\hat y}
\times
\frac{d\hat y}{dh}
\times
\frac{dh}{dW_1}
\]

Written result:

\[
(\hat y-y)\times W_2 \times x
\]

---

### Sigmoid

\[
\sigma(z)
=
\frac{1}{1+e^{-z}}
\]

Note:

- Maximum derivative ≈ 0.25
- When multiplied across many hidden layers, gradients become very small.
- This leads to vanishing gradients.

---

# First and Second Derivative

### First Derivative

- Gives slope of curve.
- If positive → function increasing.
- If negative → function decreasing.

### Second Derivative

- Concavity of curve.

Notes:

- Concave up → second derivative positive.
- Concave down → second derivative negative.

# Artificial Neural Network (ANN)

ANN = Trigonometry + Calculus (written note)

### Activation Function

Activation function introduces non-linearity into the system.

Neuron:

\[
z=x_1w_1+x_2w_2+x_3w_3+\text{Bias}
\]

Output:

\[
p=f(z)
\]

Output \(p\) becomes input to next neuron.

---

# Forward Propagation

Forward propagation is when we move from input layer to output layer through hidden layers.

Output is compared with true value to determine loss.

---

# Back Propagation

To reduce losses we perform backpropagation.

During backpropagation we change weights in hidden layers and update them in such a way that loss becomes minimum.

---

- Second derivative → change of slope (curvature).

- ReLU activation is often used to reduce vanishing-gradient problems.

- all principal components are perpendicular to eachother

- covariance Direction of the linear relationship.rer -∞ to +∞.

\[

\begin{bmatrix}
