# `Feature Scaling`
## Scaling Technique
- standarized
formula=orginal_values-mean/std

## very large variable get effect on model  thats why we do standardizationn


```python
df = pd.DataFrame({'Values':[3,6,5,1,6,8,3,8,9,4]})

df['StdScale'] = (df.Values-np.mean(df.Values))/np.std(df.Values,ddof=1)
```

```python
df
```

## Min-Max scaler
- Min-Max Scaler: (x - min(col)) / (max(col) - min(value))

- cons:this can be affected by outlier

```python
df['MinMax']=(df.Values-min(df.Values))/(max(df.Values)-min(df.Values))
df
```

## Robust Scaler :
- (x-median(col))/percentile(col,75)-percentile((col,75))

- if specially distance model like knn,lknmeans,pca this scaling down important it can be ok in regression there is no need in decission tree.

```python
df['Robust'] = (df.Values-np.median(df.Values))/(np.percentile(df.Values,75)-np.percentile(df.Values,25))
df
```
