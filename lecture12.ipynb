## One sample poisson rate

Used when we wish to compare the rate at which event occurs with a claim value and establish whether the difference is statistically significant or not.

```python
from statsmodels.stats import rates
from statsmodels.stats.rates import test_poisson
```

A good class would ask atleast 20 question everyday

h0 n>=20 ha<20 ==> left tail test

```python
questions=[14,19,24,18,17,23,21,18,19,17]
np.mean(questions)
```

```python
from statsmodels.stats.rates import test_poisson
test_poisson(sum(questions),len(questions),20,method='score',alternative='smaller')
```

## Two sample poisson rate:

- Used when we have two samples and the difference of the average counts of an event has to be compared against a claim value and establish whether the difference is statistically significant or not.

DBDA batch asks atleast 6 questions/day more than AI  batch

- Ho:avg_count(dbda)-avg(AI)>=6

- Ha:avg_count(dbda)-avg(AI)<6

- In DBDA'S 12 days of training ,145 questions are asked

- IN AI 's 5 days of traiing, 45 questions are asked

```python
from statsmodels.stats.rates import test_poisson_2indep
test_poisson_2indep(145,12,45,5,value=6,compare='diff',method='score',alternative='smaller')#questions per day diff
```

## One variance test:

- Used when the variance of a sample has to be compared against a claim value and establish whether their difference from 1 is statistically significant or not

```python
df=pd.read_excel("/content/drive/MyDrive/My pitch/CDAC_DataBook.xlsx",sheet_name="faithful")
```

```python
erupt=df.eruptions
np.var(erupt,ddof=1)
```

Ho variance(erupt) <1.1

ha variance(erupt)>1.1

ha variance (erupt) >1.1===>right tail test


```python
ts=(len(erupt)-1)*np.var(erupt,ddof=1)/1.1
```

```python
ts
from scipy.stats import chi2
p_value=1-chi2.cdf(ts,len(erupt)-1)

p_value=chi2.sf(ts,len(erupt)-1)
p_value
```

```python
np.var(erupt,ddof=1)/1.1
```

## Two variance test:

- Used when we have two samples. The ratio of the variances has to be compared against a claim value and establish whether their difference from 1 is statistically significant or not

```python
samp1=erupt[30:70]
samp2=erupt[130:180]
np.var(samp1,ddof=1)/np.var(samp2,ddof=1)
```

```python
samp_ratio=np.var(samp1,ddof=1)/np.var(samp2,ddof=1)
ts=samp_ratio/1.4
ts
```

```python
from scipy.stats import f
f.cdf(ts,len(samp1)-1,len(samp2)-1)# pvalue
```
